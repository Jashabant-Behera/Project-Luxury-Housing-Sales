```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

### Data Loading and Initial Exploration

```python
raw_data_path = '../Data/raw/Luxury_Housing_Bangalore.csv'

try:
    df = pd.read_csv(raw_data_path)
    print(f"Data loaded successfully!")
    print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
except FileNotFoundError:
    print("ERROR: File not found!")
    print("Please check the filename and path.")

# Display the dataframe
df
```

```python
print("Columns Summary:")
print(df.describe())
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
print("Dataset Info:")
print(df.info())
```

### Exploring the Dataset

```python
print("Duplicate Rows:")
duplicates = df.duplicated().sum()
print(f"Total duplicates: {duplicates}")
```

```python
### Removing Duplicates
print("Removing duplicates...")
initial_rows = len(df)
df = df.drop_duplicates()
removed = initial_rows - len(df)
print(f"Removed {removed} duplicate rows")

print(f"Duplicate Rows: {df.duplicated().sum()}")
```

```python
print(f"{df.isnull().sum()}")
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing_Count': missing,
    'Percentage': missing_percent
})
print(missing_df[missing_df['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False))
```

### Cleaning the Dataset

```python
### Cleaning Buyer_Comments...
print("Cleaning Buyer_Comments nulls...")

if 'Buyer_Comments' in df.columns:
    df['Buyer_Comments'] = df['Buyer_Comments'].replace('', np.nan)
    df['Buyer_Comments'] = df['Buyer_Comments'].str.strip()

```

```python
### Handle Amenity_Score nulls
print("Handling Amenity_Score nulls...")

print(f"Missing Values: {df['Amenity_Score'].isnull().sum()}")

if 'Amenity_Score' in df.columns:
    missing_amenity = df['Amenity_Score'].isnull().sum()
    if missing_amenity > 0:
        median_amenity = df['Amenity_Score'].median() # filling missing values with median
        df['Amenity_Score'].fillna(median_amenity, inplace=True)
        print(f"Filled {missing_amenity} nulls with median: {median_amenity}")
    else:
        print(f"No Missing Values")

print(f"Missing Values: {df['Amenity_Score'].isnull().sum()}")
```

```python
### Standardize Configuration
print(f"Standardize Configuration...")

print(f"Unique Result: {df['Configuration']}")

if "Configuration" in df.columns:
    df['Configuration'] = df['Configuration'].str.strip().str.upper()

print(f"Unique Result: {df['Configuration']}")
```

```python
### Handeling Unit_Size_Sqft column
print(f"Handeling Unit_Size_Sqft column...")

print(f"Missing Unit: {df['Unit_Size_Sqft'].isnull().sum()}")

if "Unit_Size_Sqft" in df.columns:

    negative_unit = (df['Unit_Size_Sqft'] < 0).sum()

    if negative_unit > 0:
        
        df.loc[df['Unit_Size_Sqft'] < 0, 'Unit_Size_Sqft'] = np.nan

        df['Unit_Size_Sqft'] = df.groupby('Configuration')['Unit_Size_Sqft'].transform(lambda x: x.fillna(x.median()))


    missing_unit = df['Unit_Size_Sqft'].isnull().sum()

    if missing_unit > 0:
        df['Unit_Size_Sqft'] = df.groupby('Configuration')['Unit_Size_Sqft'].transform(lambda x: x.fillna(x.median())) 

        print(f"Missing Unit filled with median according to Configuration Column")

print(f"Missing Unit: {df["Unit_Size_Sqft"].isnull().sum()}")
```

```python
### Cleaning Ticket_Price_Cr Column (FIXED VERSION)

print("Cleaning Ticket_Price_Cr column...")

print(f"\nMissing values BEFORE: {df['Ticket_Price_Cr'].isnull().sum()}")

if "Ticket_Price_Cr" in df.columns:
    
    # Replace all characters EXCEPT: digits (0-9), dot (.), and minus sign (-)
    df['Ticket_Price_Cr'] = df['Ticket_Price_Cr'].astype(str).str.replace('[^0-9.-]', '', regex=True)
    
    # Convert to numeric
    df['Ticket_Price_Cr'] = pd.to_numeric(df['Ticket_Price_Cr'], errors='coerce')
    print(f"Converted to numeric")
    
    # Check for negative values
    print("\nChecking for negative values...")
    negative_price = (df['Ticket_Price_Cr'] < 0).sum()
    print(f"Negative values found: {negative_price}")
    
    if negative_price > 0:        
        # Replace negative values with NaN
        df.loc[df['Ticket_Price_Cr'] < 0, 'Ticket_Price_Cr'] = np.nan
        print(f"Replaced {negative_price} negative values with NaN")
    
    print(f"Checking Negative values: {(df['Ticket_Price_Cr'] < 0).sum()}")

    # Fill missing values using Micro_Market + Configuration
    print("\nFilling missing values...")
    missing_before = df['Ticket_Price_Cr'].isnull().sum()
    print(f"Missing values to fill: {missing_before}")
    
    if missing_before > 0:
        # Fill by Micro_Market + Configuration
        df['Ticket_Price_Cr'] = df.groupby(['Micro_Market', 'Configuration'])['Ticket_Price_Cr'].transform(
            lambda x: x.fillna(x.median())
        )
    
    print(f"\nMissing values AFTER: {df['Ticket_Price_Cr'].isnull().sum()}")
    
```

```python
print("Standardize Micro_Market and Developer_Name...")

print(f"Unique before:\n{df[['Micro_Market', 'Developer_Name']].drop_duplicates()}")

if "Micro_Market" in df.columns:
    df['Micro_Market'] = df['Micro_Market'].str.strip().str.upper()

if "Developer_Name" in df.columns:
    df['Developer_Name'] = df['Developer_Name'].str.strip().str.upper()

print(f"Unique after:\n{df[['Micro_Market', 'Developer_Name']].drop_duplicates()}")
```

```python
print("STANDARDIZING TEXT COLUMNS")

# Transaction_Type
if 'Transaction_Type' in df.columns:
    df['Transaction_Type'] = df['Transaction_Type'].str.strip().str.title()
    print(f"\nTransaction_Type: {df['Transaction_Type'].unique()}")

# Buyer_Type
if 'Buyer_Type' in df.columns:
    df['Buyer_Type'] = df['Buyer_Type'].str.strip().str.title()
    print(f"Buyer_Type: {df['Buyer_Type'].unique()}")

# Possession_Status
if 'Possession_Status' in df.columns:
    df['Possession_Status'] = df['Possession_Status'].str.strip().str.title()
    print(f"Possession_Status: {df['Possession_Status'].unique()}")

# Sales_Channel
if 'Sales_Channel' in df.columns:
    df['Sales_Channel'] = df['Sales_Channel'].str.strip().str.title()
    print(f"Sales_Channel: {df['Sales_Channel'].unique()}")

# NRI_Buyer
if 'NRI_Buyer' in df.columns:
    df['NRI_Buyer'] = df['NRI_Buyer'].str.strip().str.title()
    print(f"NRI_Buyer: {df['NRI_Buyer'].unique()}")

print("\n All text columns standardized!")
```

```python
print("CONVERTING PURCHASE_QUARTER TO DATETIME")

if 'Purchase_Quarter' in df.columns:    
    # Convert to datetime
    df['Purchase_Quarter'] = pd.to_datetime(df['Purchase_Quarter'], errors='coerce')

    print(f"Converted to datetime!")
```
